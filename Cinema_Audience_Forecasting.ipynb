## Loading all CSVs from Kaggle

```python
import os
import pandas as pd
import numpy as np

data_dir = '/kaggle/input'  # Kaggle input path

dataframes = {}
for dirname, _, filenames in os.walk(data_dir):
    for filename in filenames:
        if filename.endswith('.csv'):
            full_path = os.path.join(dirname, filename)
            key_name = os.path.splitext(filename)[0]
            try:
                dataframes[key_name] = pd.read_csv(full_path)
                print(f"Loaded: {filename}")
            except Exception as e:
                print(f"Error reading {full_path}: {e}")
```

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import STL
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

sns.set(style="whitegrid", context="notebook", rc={"figure.figsize": (10, 5)})
plt.rcParams["figure.dpi"] = 120

SAMPLE_THEATERS_COUNT = 3  # how many theaters to sample for per-theater plots
RANDOM_STATE = 42

# Helper function
def safe_df(key):
    if key not in dataframes:
        print(f"WARNING: '{key}' not found in dataframes. Skipping related steps.")
        return None
    return dataframes[key].copy()
```

### Load the tables from Dictonary to individual dataframes

```python
print("=== Loading tables and basic normalization ===")
df_visits = safe_df("booknow_visits")
df_theaters = safe_df("booknow_theaters")
df_date = safe_df("date_info")
df_relation = safe_df("movie_theater_id_relation")
df_cine = safe_df("cinePOS_booking")
df_booking = safe_df("booknow_booking")

# Convert key datetime columns safely
if df_visits is not None and "show_date" in df_visits.columns:
    df_visits["show_date"] = pd.to_datetime(df_visits["show_date"], errors="coerce")
    df_visits.dropna(subset=["show_date"], inplace=True)
    if "audience_count" in df_visits.columns:
        df_visits["audience_count"] = pd.to_numeric(df_visits["audience_count"], errors="coerce")
    print("booknow_visits loaded:", df_visits.shape)
else:
    print("booknow_visits missing or has no show_date -> can't perform full EDA")

if df_theaters is not None:
    print("booknow_theaters loaded:", df_theaters.shape)
if df_date is not None and "show_date" in df_date.columns:
    df_date["show_date"] = pd.to_datetime(df_date["show_date"], errors="coerce")
    print("date_info loaded:", df_date.shape)
if df_relation is not None:
    print("movie_theater_id_relation loaded:", df_relation.shape)
if df_cine is not None:
    if "show_datetime" in df_cine.columns:
        df_cine["show_date"] = pd.to_datetime(df_cine["show_datetime"], errors="coerce").dt.normalize()
    if "booking_datetime" in df_cine.columns:
        df_cine["booking_date"] = pd.to_datetime(df_cine["booking_datetime"], errors="coerce").dt.normalize()
    print("cinePOS_booking loaded:", df_cine.shape)
if df_booking is not None:
    if "show_datetime" in df_booking.columns:
        df_booking["show_date"] = pd.to_datetime(df_booking["show_datetime"], errors="coerce").dt.normalize()
    if "booking_datetime" in df_booking.columns:
        df_booking["booking_date"] = pd.to_datetime(df_booking["booking_datetime"], errors="coerce").dt.normalize()
    print("booknow_booking loaded:", df_booking.shape)
```

### EDA on Feature Engineering Steps

```python
print("=== EDA on Feature Engineering Decisions ===")
df_fe = df_visits.copy()
# calendar features
df_fe["month"] = df_fe["show_date"].dt.month
df_fe["weekday"] = df_fe["show_date"].dt.weekday
df_fe["is_weekend"] = (df_fe["weekday"] >= 5).astype(int)
df_fe["show_date_ord"] = df_fe["show_date"].map(pd.Timestamp.toordinal)
```

```python
# theater historical average
theater_avg = df_fe.groupby("book_theater_id")["audience_count"].mean().rename("theater_avg")
df_fe = df_fe.merge(theater_avg.reset_index(), on="book_theater_id", how="left")
plt.figure(figsize=(8,4))
sns.histplot(theater_avg, kde=True)
plt.title("Distribution of theater_avg")
plt.show()
```

### **Interpretation for Plot: Distribution of theater_avg**

> This histogram shows the distribution of each theater’s historical average audience (`theater_avg`).
> Most theaters have average attendance in the range of **20-60**, while a smaller number of theaters exhibit much higher averages.
> This reflects long-term performance differences between theaters—some theaters consistently attract larger crowds due to factors such as location, capacity, or frequency of popular shows.
>
> Including `theater_avg` as a feature provides the model with a **stable, theater-specific baseline** that helps smooth out noisy or missing lag values.
> So, merge historical averages, especially for situations where lagged data may be absent or limited.

```python
# STL decomposition (global daily totals)
daily = df_fe.groupby("show_date")["audience_count"].sum().asfreq("D").fillna(method="ffill")
# If series too short, skip
if len(daily) >= 14:
    stl = STL(daily, period=7, robust=True).fit()
    # plot seasonality & trend
    plt.figure(figsize=(12,3))
    plt.plot(stl.seasonal)
    plt.title("STL seasonal (global daily total)")
    plt.show()
    plt.figure(figsize=(12,3))
    plt.plot(stl.trend)
    plt.title("STL trend (global daily total)")
    plt.show()

    # map components back to df_fe
    seasonal = pd.Series(stl.seasonal, index=stl.seasonal.index)
    trend = pd.Series(stl.trend, index=stl.trend.index)
    # merge by show_date
    df_fe = df_fe.merge(seasonal.rename("daily_seasonal").reset_index().rename(columns={"index":"show_date"}),
                            on="show_date", how="left")
    df_fe = df_fe.merge(trend.rename("daily_trend").reset_index().rename(columns={"index":"show_date"}),
                            on="show_date", how="left")
    # scatter seasonal/trend vs audience
    plt.figure(figsize=(7,4))
    sns.scatterplot(x="daily_seasonal", y="audience_count", data=df_fe, alpha=0.4)
    plt.title("daily_seasonal vs audience_count")
    plt.show()
    plt.figure(figsize=(7,4))
    sns.scatterplot(x="daily_trend", y="audience_count", data=df_fe, alpha=0.4)
    plt.title("daily_trend vs audience_count")
    plt.show()
else:
    print("Too few daily points for STL; skipping STL section (need >=14 days).")
```

### **Interpretation for Plot: STL seasonal (global daily total)**

> The STL decomposition's seasonal component reveals a **clear weekly repeating pattern**, with peaks and troughs occurring at regular 7-day intervals.
> This demonstrates that the overall audience demand has a strong **weekly seasonality**, with certain days of the week consistently attracting higher attendance.
> The consistent up-down oscillation across the entire timeline confirms that time-based seasonal patterns persist through different months and years.
>
> So, extract features such as `weekday`, `is_weekend`, and the include `daily_seasonal` as an additional seasonality signal for the model.

---

### **Interpretation for Plot: STL trend (global daily total)**

> The STL trend component captures long-term demand movement over time.
> The plot shows a **gradual rise in total audience turnout**, followed by fluctuations that correspond to broader seasonal events-such as festivals, holiday periods, or high-demand movie releases.
> This long-horizon trend provides the model with insight into **overall demand growth or decline**, beyond daily fluctuations.
>
> Include `daily_trend`, which helps the model understand long-term shifts in audience behavior and improves prediction accuracy during upward or downward demand phases.

---

### **Interpretation for Plot: daily_seasonal vs audience_count**

> This scatterplot compares the STL seasonal component (`daily_seasonal`) with actual audience_count.
> Although the relationship is not strictly linear, shows with higher `daily_seasonal` values generally tend to have larger audiences.
> This indicates that the STL seasonal component captures meaningful cyclical demand behavior, especially for high-volume days.
>
> So, extract `daily_seasonal` value as a feature, since it provides a model-friendly representation of weekly cyclicity that the raw date alone cannot capture.

---

### **Interpretation for Plot: daily_trend vs audience_count**

> This plot shows how audience_count varies with the STL trend component.
> A broad positive association is visible: periods with higher trend values often correspond to higher audience turnout for many shows.
> This demonstrates that the trend component successfully captures **macro-level demand shifts** - such as season changes or holiday surges.
>
> Including `daily_trend` allows the model to learn from long-term demand changes rather than only short-term variations.

```python
# Lag features (per-theater)
df_fe = df_fe.sort_values(["book_theater_id","show_date"])
df_fe["lag_1"] = df_fe.groupby("book_theater_id")["audience_count"].shift(1)
df_fe["lag_7"] = df_fe.groupby("book_theater_id")["audience_count"].shift(7)
df_fe["rolling_7"] = (df_fe.groupby("book_theater_id")["audience_count"]
                         .shift(1).rolling(7, min_periods=1).mean().reset_index(level=0, drop=True))

# fill missing with theater_avg
df_fe["lag_1"] = df_fe["lag_1"].fillna(df_fe["theater_avg"])
df_fe["lag_7"] = df_fe["lag_7"].fillna(df_fe["theater_avg"])
df_fe["rolling_7"] = df_fe["rolling_7"].fillna(df_fe["theater_avg"])
```

```python
# ACF / PACF on global daily for lag justification
plt.figure(figsize=(10,2))
plot_acf(daily, lags=30, zero=False)
plt.title("ACF (global daily totals)")
plt.show()
plt.figure(figsize=(10,2))
plot_pacf(daily, lags=30, zero=False, method="ywm")
plt.title("PACF (global daily totals)")
plt.show()
```

### **Interpretation for Plot: ACF (global daily totals)**

> The ACF plot shows strong positive autocorrelation at multiple lag values-especially at lag 1, lag 7, and subsequent multiples of 7.
> This clearly confirms **weekly seasonality** in audience demand.
> The persistent autocorrelation at lag 1 and lag 7 provides strong evidence that past audience values are highly predictive of future values.
>
> So, create `lag_1`, `lag_7`, and rolling windows as part of the feature engineering process.

---

### **Interpretation for Plot: PACF (global daily totals)**

> The PACF plot reveals strong partial correlations at lag 1 and lag 7, while most other lags fall within confidence intervals.
> This indicates that the audience_count is primarily influenced by the previous day (lag 1) and the previous week (lag 7), with limited independent contributions from other lags.
>
> So, use **lag_1 and lag_7** specifically, rather than many higher-order lags, which keeps the model simpler and reduces noise.

## Data Preprocessing

```python
import gc, warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
from datetime import datetime

from statsmodels.tsa.seasonal import STL
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import RandomForestRegressor
from catboost import CatBoostRegressor, Pool
import lightgbm as lgb
import xgboost as xgb

RANDOM_STATE = 42
```

```python
# 1) LOAD PRIMARY TARGET TABLE
df = dataframes["booknow_visits"].copy()
df["show_date"] = pd.to_datetime(df["show_date"], errors="coerce")
df["audience_count"] = df["audience_count"].astype(float)
df.dropna(subset=["show_date"], inplace=True)

print("Initial df shape:", df.shape)

# 2) MERGE BOOKNOW THEATER INFO
booknow_th = dataframes["booknow_theaters"][["book_theater_id","theater_type","theater_area"]]
df = df.merge(booknow_th, on="book_theater_id", how="left")

# 3) MERGE DATE INFO
date_info = dataframes["date_info"].copy()
date_info["show_date"] = pd.to_datetime(date_info["show_date"])
df = df.merge(date_info, on="show_date", how="left")

# 4) MERGE THEATER RELATION (MAP TO CINE THEATER)
print("Before relation merge:", df.shape)
relation = dataframes["movie_theater_id_relation"].copy()
df = df.merge(relation, on="book_theater_id", how="left")

print("After relation merge:", df.shape)

# 5) CINEPOS AGGREGATION (FEATURE)
cine = dataframes["cinePOS_booking"].copy()
cine["show_date"] = pd.to_datetime(cine["show_datetime"]).dt.normalize()

cine_agg = (
    cine.groupby(["cine_theater_id", "show_date"])["tickets_sold"]
    .sum().reset_index().rename(columns={"tickets_sold": "cine_tickets_sold"})
)

df = df.merge(cine_agg, on=["cine_theater_id","show_date"], how="left")
df["cine_tickets_sold"] = df["cine_tickets_sold"].fillna(0)

# 6) BOOKNOW BOOKING AGGREGATION (FEATURE)
bbook = dataframes["booknow_booking"].copy()
bbook["show_date"] = pd.to_datetime(bbook["show_datetime"]).dt.normalize()

bbook_agg = (
    bbook.groupby(["book_theater_id","show_date"])["tickets_booked"]
    .sum().reset_index()
)

df = df.merge(bbook_agg, on=["book_theater_id","show_date"], how="left")
df["tickets_booked"] = df["tickets_booked"].fillna(0)

print("After CinePOS + BookNow booking merges:", df.shape)

# 7) CALENDAR FEATURES
df["month"] = df["show_date"].dt.month
df["weekday"] = df["show_date"].dt.weekday
df["is_weekend"] = (df["weekday"] >= 5).astype(int)
df["show_date_ord"] = df["show_date"].map(datetime.toordinal)

# 8) THEATER HISTORICAL AVERAGES (FEATURE)
theater_avg = df.groupby("book_theater_id")["audience_count"].mean().rename("theater_avg")
df = df.merge(theater_avg.reset_index(), on="book_theater_id", how="left")

# 9) STL DECOMPOSITION
daily = df.groupby("show_date")["audience_count"].sum().asfreq("D").fillna(method="ffill")
stl = STL(daily, period=7).fit()
seasonal = stl.seasonal
trend = stl.trend

df["daily_seasonal"] = df["show_date"].map(seasonal).fillna(0)
df["daily_trend"] = df["show_date"].map(trend).fillna(0)

# 10) LAG FEATURES (PER THEATER)
df = df.sort_values(["book_theater_id","show_date"])

df["lag_1"] = df.groupby("book_theater_id")["audience_count"].shift(1)
df["lag_7"] = df.groupby("book_theater_id")["audience_count"].shift(7)
df["rolling_7"] = df.groupby("book_theater_id")["audience_count"].shift(1).rolling(7,min_periods=1).mean()

for c in ["lag_1","lag_7","rolling_7"]:
    df[c] = df[c].fillna(df["theater_avg"])

# 11) MODELING DATASET
model_df = df.dropna(subset=["audience_count"]).copy()

cat_cols = ["book_theater_id","theater_type","theater_area"]
num_cols = [
    "cine_tickets_sold","tickets_booked","lag_1","lag_7","rolling_7",
    "month","weekday","is_weekend","show_date_ord",
    "daily_seasonal","daily_trend","theater_avg"
]

X = model_df[cat_cols + num_cols].copy()
y = model_df["audience_count"].copy()
dates = model_df["show_date"].copy()
print("COMPLITION OF DATA PREPROCESSING")
```

```python
# 12) TIME SERIES SPLIT
tscv = TimeSeriesSplit(n_splits=3)
train_idx, valid_idx = list(tscv.split(X))[-1]

X_train, X_valid = X.iloc[train_idx], X.iloc[valid_idx]
y_train, y_valid = y.iloc[train_idx], y.iloc[valid_idx]
```

```python
# 13) LABEL ENCODING FOR LGB/XGB
le_map = {}
X_train_enc = X_train.copy()
X_valid_enc = X_valid.copy()

for c in cat_cols:
    le = LabelEncoder()
    le.fit(pd.concat([X_train[c], X_valid[c]]).astype(str))
    X_train_enc[c] = le.transform(X_train[c].astype(str))
    X_valid_enc[c] = le.transform(X_valid[c].astype(str))
    le_map[c] = le
```

## Model Training

```python
# CATBOOST MODEL
X_train_cb = X_train.copy()
X_valid_cb = X_valid.copy()

# Null imputation
for c in cat_cols:
    X_train_cb[c] = X_train_cb[c].astype(str).fillna("Unknown")
    X_valid_cb[c] = X_valid_cb[c].astype(str).fillna("Unknown")

# Create pool for traing
train_pool = Pool(X_train_cb, y_train, cat_features=cat_cols)
valid_pool = Pool(X_valid_cb, y_valid, cat_features=cat_cols)


cb = CatBoostRegressor(
    depth=6, iterations=250, learning_rate=0.05,
    random_seed=RANDOM_STATE, loss_function="RMSE",
    early_stopping_rounds=50, verbose=False
)
cb.fit(train_pool, eval_set=valid_pool)

# LIGHTGBM MODEL
lgb_train = lgb.Dataset(X_train_enc, label=y_train)
lgb_valid = lgb.Dataset(X_valid_enc, label=y_valid)

lgb_params = {
    "objective": "regression",
    "metric": "rmse",
    "learning_rate": 0.05,
    "num_leaves": 31,
    "min_child_samples": 20
}

# Training of model
lgb_model = lgb.train(
    lgb_params,
    lgb_train,
    valid_sets=[lgb_valid],
    num_boost_round=500,
    callbacks=[lgb.early_stopping(stopping_rounds=50)]
)

# XGBOOST MODEL
dtrain = xgb.DMatrix(X_train_enc, label=y_train)
dvalid = xgb.DMatrix(X_valid_enc, label=y_valid)

xgb_params = {
    "objective": "reg:squarederror",
    "eta": 0.05,
    "max_depth": 6
}

# Training of model
xgb_model = xgb.train(
    xgb_params,
    dtrain,
    evals=[(dvalid,"valid")],
    num_boost_round=600,
    early_stopping_rounds=50,
    verbose_eval=False
)

# RANDOM FOREST MODEL
rf = RandomForestRegressor(
    n_estimators=400,
    max_depth=12,
    random_state=RANDOM_STATE,
    n_jobs=-1
)
# Training of model
rf.fit(X_train_enc, y_train)

# VALIDATION SCORES

# Predictions
pred_cb = cb.predict(X_valid_cb)
pred_lgb = lgb_model.predict(X_valid_enc)
pred_xgb = xgb_model.predict(xgb.DMatrix(X_valid_enc))
pred_rf = rf.predict(X_valid_enc)

# RMSE
rmse_cb = np.sqrt(mean_squared_error(y_valid, pred_cb))
rmse_lgb = np.sqrt(mean_squared_error(y_valid, pred_lgb))
rmse_xgb = np.sqrt(mean_squared_error(y_valid, pred_xgb))
rmse_rf = np.sqrt(mean_squared_error(y_valid, pred_rf))

# R² Scores
r2_cb = r2_score(y_valid, pred_cb)
r2_lgb = r2_score(y_valid, pred_lgb)
r2_xgb = r2_score(y_valid, pred_xgb)
r2_rf = r2_score(y_valid, pred_rf)

print("Validation Metrics")
print(f"CatBoost => RMSE: {rmse_cb:.4f} & R2: {r2_cb:.4f}")
print(f"LightGBM => RMSE: {rmse_lgb:.4f} & R2: {r2_lgb:.4f}")
print(f"XGBoost => RMSE: {rmse_xgb:.4f} & R2: {r2_xgb:.4f}")
print(f"RandomForest => RMSE: {rmse_rf:.4f} & R2: {r2_rf:.4f}")
```

```python
RUN_TUNING = True  # set False to skip tuning and use defaults

# CATBOOST MODEL + Grid Search
X_train_cb = X_train.copy()
X_valid_cb = X_valid.copy()

for c in cat_cols:
    X_train_cb[c] = X_train_cb[c].astype(str).fillna("Unknown")
    X_valid_cb[c] = X_valid_cb[c].astype(str).fillna("Unknown")

train_pool = Pool(X_train_cb, y_train, cat_features=cat_cols)
valid_pool = Pool(X_valid_cb, y_valid, cat_features=cat_cols)

# Default / fallback catboost params
cb_best_params = {"depth": 6, "learning_rate": 0.05, "l2_leaf_reg": 3}
cb_best_iteration = 250

if RUN_TUNING:
    print("CatBoost manual grid search...")
    # Parameter Grid
    depths = [4, 6, 8, 10]
    lrs = [0.02, 0.05, 0.08]
    l2s = [1, 3, 5]
    best_rmse = float("inf")

    for depth in depths:
        for lr in lrs:
            for l2 in l2s:
                params = {
                    "depth": depth,
                    "learning_rate": lr,
                    "l2_leaf_reg": l2,
                    "iterations": 300,
                    "random_seed": RANDOM_STATE,
                    "loss_function": "RMSE"
                }
                model_tmp = CatBoostRegressor(**params, verbose=False)
                # early stopping with eval_set
                model_tmp.fit(train_pool, eval_set=valid_pool, early_stopping_rounds=50, verbose=False)

                # Predict and validate
                preds = model_tmp.predict(X_valid_cb)
                rmse = np.sqrt(mean_squared_error(y_valid, preds))
                print(f"CB depth={depth} lr={lr} l2={l2} -> RMSE: {rmse:.4f}")
                if rmse < best_rmse:
                    best_rmse = rmse
                    cb_best_params = {"depth": depth, "learning_rate": lr, "l2_leaf_reg": l2}
                    # capture best iteration if available
                    try:
                        bi = model_tmp.get_best_iteration()
                        cb_best_iteration = bi if bi and bi > 0 else params["iterations"]
                    except Exception:
                        cb_best_iteration = params["iterations"]
    print("Best CatBoost params:", cb_best_params, "best_iter:", cb_best_iteration)
else:
    print("Skipping CatBoost tuning, using defaults.")

# Train primary CatBoost model
cb = CatBoostRegressor(
    depth=cb_best_params["depth"],
    iterations=250,
    learning_rate=cb_best_params["learning_rate"],
    l2_leaf_reg=cb_best_params.get("l2_leaf_reg", 3),
    random_seed=RANDOM_STATE,
    loss_function="RMSE",
    early_stopping_rounds=50,
    verbose=False
)
cb.fit(train_pool, eval_set=valid_pool)
```

```python
# LIGHTGBM MODEL + Grid Search
lgb_train = lgb.Dataset(X_train_enc, label=y_train)
lgb_valid = lgb.Dataset(X_valid_enc, label=y_valid)

# Default lgb params
lgb_params = {
    "objective": "regression",
    "metric": "rmse",
    "learning_rate": 0.05,
    "num_leaves": 31,
    "min_child_samples": 20
}
lgb_best_params = lgb_params.copy()
lgb_best_iteration = 500

if RUN_TUNING:
    print("LightGBM manual grid search...")
    # Parameter Grid
    lr_list = [0.02, 0.05, 0.08]
    num_leaves_list = [31, 50, 70]
    min_child_list = [10, 20, 30]
    best_rmse = float("inf")

    for lr in lr_list:
        for nl in num_leaves_list:
            for mcs in min_child_list:
                params_tmp = {
                    "objective": "regression",
                    "metric": "rmse",
                    "learning_rate": lr,
                    "num_leaves": nl,
                    "min_child_samples": mcs,
                    "verbosity": -1
                }
                model_tmp = lgb.train(
                    params_tmp,
                    lgb_train,
                    valid_sets=[lgb_valid],
                    num_boost_round=400,
                    callbacks=[lgb.early_stopping(stopping_rounds=30)],
                )

                # Predict and Validate
                preds = model_tmp.predict(X_valid_enc)
                rmse = np.sqrt(mean_squared_error(y_valid, preds))
                print(f"LGB lr={lr} num_leaves={nl} min_child={mcs} -> RMSE: {rmse:.4f}")
                if rmse < best_rmse:
                    best_rmse = rmse
                    lgb_best_params = params_tmp
                    try:
                        lgb_best_iteration = model_tmp.best_iteration
                    except Exception:
                        lgb_best_iteration = 400
    print("Best LightGBM params:", lgb_best_params, "best_iter:", lgb_best_iteration)
else:
    print("Skipping LightGBM tuning, using defaults.")

# Train primary LightGBM model
lgb_model = lgb.train(
    lgb_best_params,
    lgb_train,
    valid_sets=[lgb_valid],
    num_boost_round=500,
    callbacks=[lgb.early_stopping(stopping_rounds=50)]
)
```

## Ensembling and Retraining

```python
# ENSEMBLING OF THREE MODELS
# inverse-RMSE weights
inv = np.array([1/rmse_cb, 1/rmse_lgb, 1/rmse_xgb])
weights = inv / inv.sum()
print("Ensemble weights:", weights)

# FINAL ENSEMBLE PREDICTION ON VALIDATION
pred_ensemble_valid = (
    weights[0] * pred_cb +
    weights[1] * pred_lgb +
    weights[2] * pred_xgb
)

# R2 SCORE
r2_ensemble = r2_score(y_valid, pred_ensemble_valid)
print("Final Ensemble Validation R2:", r2_ensemble)

# REFIT ON FULL TRAINING SET
# CatBoost retrain on full set using cb_best params and cb_best_iteration captured during tuning
X_full_cb = X.copy()
for c in cat_cols:
    X_full_cb[c] = X_full_cb[c].astype(str).fillna("Unknown")
full_pool = Pool(X_full_cb, y, cat_features=cat_cols)

cb_final_iters = cb_best_iteration if cb_best_iteration is not None else 250
cb_final = CatBoostRegressor(
    depth=cb_best_params["depth"],
    iterations=cb_final_iters,
    learning_rate=cb_best_params["learning_rate"],
    l2_leaf_reg=cb_best_params.get("l2_leaf_reg", 3),
    random_seed=RANDOM_STATE,
    verbose=False
)
cb_final.fit(full_pool)

# LGB full retrain using lgb_best_params and iteration count
X_full_enc = X.copy()
for c in cat_cols:
    X_full_enc[c] = le_map[c].transform(X_full_enc[c].astype(str))

lgb_train_full = lgb.Dataset(X_full_enc, label=y)
lgb_iters = lgb_best_iteration if lgb_best_iteration is not None else 500

lgb_final = lgb.train(
    lgb_best_params,
    lgb_train_full,
    num_boost_round=lgb_iters
)

# XGB full retrain using xgb_best_params and iteration count
dtrain_full = xgb.DMatrix(X_full_enc, label=y)
xgb_iters = xgb_best_iteration if xgb_best_iteration is not None else 600

xgb_final = xgb.train(
    xgb_best_params,
    dtrain_full,
    num_boost_round=xgb_iters
)

# CATBOOST FINAL MODEL VALIDATION R2
pred_cb_f = cb_final.predict(X_valid_cb)
r2_cb_f = r2_score(y_valid, pred_cb_f)
print("Final CATBOOST R2:", r2_cb_f)

# LGB FINAL
pred_lgb_f = lgb_final.predict(X_valid_enc)
r2_lgb_f = r2_score(y_valid, pred_lgb_f)
print("Final LGB R2:", r2_lgb_f)

# XGB FINAL
pred_xgb_f = xgb_final.predict(xgb.DMatrix(X_valid_enc))
r2_xgb_f = r2_score(y_valid, pred_xgb_f)
print("Final XGB R2:", r2_xgb_f)

# ENSEMBLE FINAL R2
pred_ens_f = (
    weights[0] * pred_cb_f +
    weights[1] * pred_lgb_f +
    weights[2] * pred_xgb_f
)
r2_en_f = r2_score(y_valid, pred_ens_f)
print("Final Ensemble (Full Retrain) R2:", r2_en_f)
```

```python
# PREPARE TEST DATAFRAME
sub = dataframes["sample_submission"].copy()
tmp = sub["ID"].str.rsplit("_", n=1, expand=True)
sub["book_theater_id"] = tmp[0]
sub["show_date"] = pd.to_datetime(tmp[1], errors="coerce")

# attach static info
static = df[["book_theater_id","theater_type","theater_area","theater_avg"]].drop_duplicates()
sub = sub.merge(static, on="book_theater_id", how="left")

# map STL components
sub["daily_seasonal"] = sub["show_date"].map(seasonal).fillna(0)
sub["daily_trend"] = sub["show_date"].map(trend).fillna(0)

# calendar
sub["month"] = sub["show_date"].dt.month
sub["weekday"] = sub["show_date"].dt.weekday
sub["is_weekend"] = (sub["weekday"]>=5).astype(int)
sub["show_date_ord"] = sub["show_date"].map(datetime.toordinal)

# lags: use last known values
history = df.sort_values(["book_theater_id","show_date"]).groupby("book_theater_id")["audience_count"].apply(list)

def get_lags(row):
    hist = history.get(row["book_theater_id"], [])
    if len(hist)==0:
        return pd.Series({"lag_1":0,"lag_7":0,"rolling_7":0})
    lag1 = hist[-1]
    lag7 = hist[-7] if len(hist)>=7 else np.mean(hist)
    roll7 = np.mean(hist[-7:])
    return pd.Series({"lag_1":lag1,"lag_7":lag7,"rolling_7":roll7})

sub = pd.concat([sub, sub.apply(get_lags, axis=1)], axis=1)

# ensure num_cols present
for c in num_cols:
    if c not in sub.columns:
        sub[c] = 0

# MAKE FINAL PREDICTIONS
# prepare
X_sub = sub[cat_cols + num_cols].copy()

# catboost
X_sub_cb = X_sub.copy()
for c in cat_cols:
    X_sub_cb[c] = X_sub_cb[c].astype(str)
pred_cb_sub = cb_final.predict(X_sub_cb)

# lgb/xgb encoded
X_sub_enc = X_sub.copy()

for c in cat_cols:
    X_sub_enc[c] = X_sub_enc[c].astype(str).apply(
        lambda x: le_map[c].transform([x])[0] if x in le_map[c].classes_ else -1
    )
pred_lgb_sub = lgb_final.predict(X_sub_enc)
pred_xgb_sub = xgb_final.predict(xgb.DMatrix(X_sub_enc))

final_sub = (
    weights[0]*pred_cb_sub +
    weights[1]*pred_lgb_sub +
    weights[2]*pred_xgb_sub
)

sub["audience_count"] = np.clip(np.round(final_sub),0,None).astype(int)

# Save the submission
sub[["ID","audience_count"]].to_csv("submission.csv", index=False)
print("Saved submission.csv")
# print(sub.head())
```
